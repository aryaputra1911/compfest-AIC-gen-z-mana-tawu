```python
import pandas as pd
import os

RAW_PATH = "/content/drive/MyDrive/Reflow/data/raw/FreshRetailNet-50K/data"

TRAIN_PATH = os.path.join(RAW_PATH, "train.parquet")
EVAL_PATH = os.path.join(RAW_PATH, "eval.parquet")

print("Train:", TRAIN_PATH)
print("Eval :", EVAL_PATH)
```

```python
train_df = pd.read_parquet(TRAIN_PATH)
eval_df = pd.read_parquet(EVAL_PATH)

print("Train shape:", train_df.shape)
print("Eval shape:", eval_df.shape)
```

```python
def validate_dataset(df, name):
    print(f"===== {name} =====")
    print("Rows:", len(df))
    print("Unique store_id:", df["store_id"].nunique())
    print("Unique product_id:", df["product_id"].nunique())
    print("Unique first_category_id:", df["first_category_id"].nunique())
    print("Date range:", df["dt"].min(), "-", df["dt"].max())
    print("\nNull counts per column:")
    print(df.isnull().sum())
    print("\nArray length check (hours_sale):")
    print(df["hours_sale"].apply(len).value_counts())
    print("\nArray length check (hours_stock_status):")
    print(df["hours_stock_status"].apply(len).value_counts())
    print()

validate_dataset(train_df, "TRAIN")
validate_dataset(eval_df, "EVAL")
```

```python
import numpy as np

# Stack semua array hours_sale jadi matrix (rows x 24)
sales_matrix = np.stack(train_df["hours_sale"].values)

# Rata-rata sales per jam, across seluruh dataset
avg_sales_per_hour = sales_matrix.mean(axis=0)

for hour, val in enumerate(avg_sales_per_hour):
    print(f"Jam {hour:02d}: avg_sales = {val:.5f}")
```

```python
stock_matrix = np.stack(train_df["hours_stock_status"].values)
avg_stockout_per_hour = stock_matrix.mean(axis=0)

for hour, val in enumerate(avg_stockout_per_hour):
    print(f"Jam {hour:02d}: avg_stockout_rate = {val:.4f}")
```
